```python

# SELF ORGANIZING MAP (SOM)

# Training samples from the notebook
training_data = [
    [1, 0, 1, 0],
    [1, 0, 0, 0],
    [1, 1, 1, 1],
    [0, 1, 1, 0]
]

# Learning rate
eta = 0.6

# Number of units
units = 2

# Number of input features
dimensions = 4


# Get number of epochs from user
epochs = int(input("Enter number of epochs: "))


# Get initial weight matrix from user
weights = []

print("\nEnter initial weight matrix:")

for i in range(units):

    row = list(map(float, input(
        f"Enter weights of Unit {i+1}: "
    ).split()))

    if len(row) != dimensions:
        raise ValueError("Enter exactly 4 weights per unit")

    weights.append(row)


# Function to calculate Squared Euclidean Distance
def squared_euclidean_distance(x, w):

    distance = 0

    for j in range(len(x)):
        distance += (x[j] - w[j]) ** 2

    return distance


# Display initial weight matrix
print("\nInitial Weight Matrix:")

for i in range(units):
    print(f"Unit {i+1}: {weights[i]}")


# START TRAINING
for epoch in range(epochs):

    print("\n")
    print("=" * 60)
    print(f"                  EPOCH {epoch+1}")
    print("=" * 60)

    # Process each training sample
    for sample_no, x in enumerate(training_data):

        print(f"\nTraining Sample {sample_no+1}: {x}")

        # STEP 1: Calculate squared Euclidean distances
        distances = []

        for i in range(units):

            d = squared_euclidean_distance(x, weights[i])

            distances.append(d)

            print(f"Distance from Unit {i+1} = {d:.4f}")

        # STEP 2: Find winning unit
        winner = distances.index(min(distances))

        print(f"\nWinning Unit = Unit {winner+1}")

        # STEP 3: Update winning unit weights
        print("\nWeight Updation:")

        old_weights = weights[winner].copy()

        for j in range(dimensions):

            weights[winner][j] = (
                weights[winner][j]
                + eta * (x[j] - weights[winner][j])
            )

        print("Old weights:", old_weights)

        print("New weights:", weights[winner])

        # STEP 4: Display updated weight matrix
        print("\nUpdated Weight Matrix:")

        for i in range(units):

            print(f"Unit {i+1}: {weights[i]}")

    # Display weights after each epoch
    print("\n" + "-" * 60)
    print(f"WEIGHT MATRIX AFTER EPOCH {epoch+1}")
    print("-" * 60)

    for i in range(units):

        print(f"Unit {i+1}: {weights[i]}")


# FINAL RESULT
print("\n")
print("=" * 60)
print("              FINAL WEIGHT MATRIX")
print("=" * 60)

for i in range(units):

    print(f"Unit {i+1}: {weights[i]}")

print("\nTraining Completed!")
```
